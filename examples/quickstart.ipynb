# Create a Sketch with ipysketch_lite 🎨

**ipysketch_lite** is a Python library for creating interactive sketches in Jupyter notebooks.

## Key features

- 👨‍🎨 Sketch drawings directly inside your notebook.
- 🔧 Tools for common drawing utilities.
- 💻 Sketch data is directly available in the Jupyter notebook. 
- 🏃‍♂️ Run in **VSCode**, **Jupyter**, **JupyterLab**, ...


The following is an example of how to start sketching in a Jupyter notebook.

First install the package

```python
%pip install ipysketch_lite
```

```python
from ipysketch_lite import Sketch

sketch = Sketch()
```

Draw in the widget before running the cells below. After each edit, rerun
them to retrieve the latest drawing.

- `sketch.image`: a PIL image.
- `sketch.data`: a PNG data URL.
- `sketch.svg`: SVG text, including stroke widths and bucket fills.

Use `RasterSketch` for the original raster-only widget.

```python
# Print the first 100 characters of the image data
print(f"Sketch base64 image data: {sketch.data[:100]}...")

sketch.image
```

```python
from IPython.display import SVG, display

display(SVG(sketch.svg))
```
