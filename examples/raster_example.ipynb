# Raster Sketch examples 📝

Use `RasterSketch` for the original canvas-based drawing widget.
It supports brush size, color, bucket fill, undo, and redo.
Draw in the widget, then run the following cells to retrieve the image.
Raster sketches expose `.image`, `.data` To sketch over an existing `Image` use the `AnnotationSketch`.

First generating a sample image to annotate over. The following plot contains stats that could use annotations.

```python
import io
import matplotlib.pyplot as plt
from PIL import Image

names = ["Alice", "Bob", "Charlie", "David", "Emma"]
values = [9, 15, 7, 12, 20]

fig, ax = plt.subplots()
ax.bar(names, values, color="skyblue")
ax.plot(names, values, color="red")
ax.set_title("Example plot")

buf = io.BytesIO()
fig.savefig(buf, format="png")
plt.close(fig)
pil_image = Image.open(buf)
```

Pass the `pil_image: Image` as an argument to `AnnotationSketch` to start drawing over it

```python
from ipysketch_lite import AnnotationSketch

sketch = AnnotationSketch(pil_image)
```

Annotate the image in the widget, then run the next cell to retrieve it.
Rerun the cell after further edits.
